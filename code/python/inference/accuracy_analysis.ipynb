```python
import numpy as np
import pandas as pd
```

## Load Ground Truth and Data

```python
df_gt = pd.read_csv("../../../data/stimuli/inference/drop_gt.csv").drop(columns=["Unnamed: 0"])
# df_gt hole is zero indexed, so add 1 to make it one indexed
df_gt["hole"] = df_gt["hole"] + 1
df_gt.rename(columns={"hole": "gt_hole"}, inplace=True)

df_judge_vision = pd.read_csv("../../../data/human_data/inference/vision_judgment_rt_cleaned.csv").drop(columns=["Unnamed: 0"])
df_judge_audio = pd.read_csv("../../../data/human_data/inference/audio_judgment_rt_cleaned.csv").drop(columns=["Unnamed: 0"])
df_judge_occluded = pd.read_csv("../../../data/human_data/inference/occluded_judgment_rt_cleaned.csv").drop(columns=["Unnamed: 0"])
```

```python
def calculate_percent_correct(df_human, df_gt=df_gt):
    df_merged = pd.merge(df_human, df_gt, on="trial", how="inner")
    df_merged["correct"] = df_merged["response"] == df_merged["gt_hole"]
    return np.mean(df_merged["correct"])
```

```python
print("Vision", np.round(calculate_percent_correct(df_judge_vision), 2))
print("Audio", np.round(calculate_percent_correct(df_judge_audio), 2))
print("Occluded", np.round(calculate_percent_correct(df_judge_occluded), 2))
```
